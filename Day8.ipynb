```python
x = np.arange(5)
print(x)

fig, axs = plt.subplots(1,2)

axs[0].plot(x,x**2,marker='o',markerfacecolor='red')
axs[0].set_title("Square")
axs[0].set_xlabel("Numbers")
axs[0].set_ylabel("Square of numbers")

axs[1].plot(x,x**3,marker='s',markerfacecolor='red')
axs[1].set_title("Cube")
axs[1].set_xlabel("Numbers")
axs[1].set_ylabel("Cube of numbers")

plt.tight_layout()
plt.show()
```

```python
plt.style.use('dark_background')
days = np.arange(7)

apple = [150, 152, 153, 151, 155, 157, 156]
google = [2800, 2820, 2810, 2830, 2840, 2850, 2860]
amazon = [3400, 3420, 3410, 3430, 3440, 3450, 3460]
microsoft = [299, 301, 300, 302, 303, 305, 304]

fig, axs = plt.subplots(2,2,figsize=(10,8))

axs[0,0].plot(days, apple , marker='o' , color = 'red')
axs[0,0].set_title("Apple Stock Price")
axs[0,0].grid(True)

axs[0, 1].plot(days, google, marker='s', color='blue')
axs[0, 1].set_title("Google Stock Price")
axs[0, 1].grid(True)

axs[1, 0].plot(days, amazon, marker='^', color='green')
axs[1, 0].set_title("Amazon Stock Price")
axs[1, 0].grid(True)

axs[1, 1].plot(days, microsoft, marker='d', color='purple')
axs[1, 1].set_title("Microsoft Stock Price")
axs[1, 1].grid(True)

for ax in axs[1, :]:
    ax.set_xlabel("Day")
for ax in axs[:, 0]:
    ax.set_ylabel("Price")

plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(2, 1,sharex=True)

axs[0].plot([1,2,3], [3,2,1])
axs[1].plot([1,2,3], [1,2,3])

plt.show()
```

```python
x = np.arange(10)
y = x**2
print(y)


plt.plot(x, y, linestyle=':', marker='o',color='green',markerfacecolor='red',linewidth=2,label='Square of Numbers')

plt.title('Customized Plot', fontsize=20)

plt.xlabel('X-axis', rotation=90)
plt.ylabel('Y-axis',rotation=45)

plt.grid()
plt.xticks(x)
#plt.yticks(y)
plt.legend()
plt.show()
```

```python
days = np.arange(5)  # Day 0,1,2,3,4

organic = [50, 60, 70, 80, 90]      
social = [30, 40, 50, 60, 30]        
referral = [20, 25, 15, 20, 25]      

plt.stackplot(days, organic, social, referral, labels=['Organic', 'Social', 'Referral'])

plt.title("Website Traffic by Source")
plt.xlabel("Day")
plt.ylabel("Number of Visitors")
#plt.legend()
plt.legend(loc='upper right')

plt.show()
```
