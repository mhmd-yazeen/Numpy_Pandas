```python
profit = [
    12, 15, 18, 20, 22, 19, 25, 24, 21, 23,27, 29, 31, 28, 26, 30, 32, 35, 120
]
```

```python
sns.boxplot(x=profit)
plt.title('Monthly Profit')
plt.xlabel('Profit')
plt.show()
```

```python
q1 = np.percentile(profit, 25)
q3 = np.percentile(profit, 75)
iqr = q3 - q1
print(f"Q1: {q1}")
print(f"Q3: {q3}")
print(f"IQR: {iqr}")
```

```python
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")
```

```python
outliers=[x for x in profit if x<lower_bound or x>upper_bound]
print(outliers)
```

```python
q3=np.percentile(sales,25)
q4=np.percentile(sales,75)
iqr=q4-q3

print(f'Q1: {q3}')
print(f'Q2: {q4}')
```

```python
lower = q3 - 1.5*iqr
upper = q4 + 1.5*iqr

print(f'Lower bound: {lower}')
print(f'Upper bound: {upper}')
```

```python
outliers=[x for x in sales if x<lower_bound or x>upper_bound]
print(outliers)
```
