```python
df['year_added'] = df['date_added'].dt.year   
df['month_added'] = df['date_added'].dt.month 
df['day_added'] = df['date_added'].dt.day     
```

```python
def extract_duration(x):
    if pd.isna(x):
        return np.nan
    return int(x.split()[0])

df['duration_int'] = df["duration"].apply(extract_duration)
```
